```python
#a. When does the code in a function execute: when the function is defined or called?

#The code executes when the function is called, not when the function is first defined.
```

```python
#b. How many global scopes are there in a Python program? How many local scopes?

#There is only one global scope, created when program begins. When program terminates, it destroys the global scope, and all of its variables get forgotten.
#A new local scope gets created whenever a program calls a function. So how many local scopes depends on how many funtions there are in a python program
```

```python
#c. What happens to variables in a local scope when the function call returns?

#When the function returns, the local scope gets destroyed, along with all variables assigned in the function.
```

```python
#d. How can you prevent a program from crashing when it gets an error?

#Errors can be handled with try and except statements. The code that could potentially have an error is put in a try clause.
#The program execution moves to the start of a following except clause if an error happens.
#When code in a try clause causes an error, the program execution immediately moves to the code in the except clause.
#After running that code, the execution continues as normal. If the program doesn’t raise an exception in the try clause, the program skips the code in the except clause.
#Therefore the program won't crash
```

```python
#Practice Program The Collatz Sequence

def collatz(num):
    if num % 2 == 0:# even
        result = num // 2
        return result
    else:
        result = 3 * num + 1# odd
        return result
print("Enter number：")

try:
  number = int(input())
  all_num = []
  while number != 1:
    all_num.append(number)
    number = collatz(number)
  all_num.append(1)
  print(all_num, sep=' ')

except ValueError:
  print("Please enter an integer number")
```
