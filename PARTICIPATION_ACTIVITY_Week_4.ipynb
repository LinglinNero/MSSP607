```python
from google.colab import drive
drive.mount('/content/drive')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rc('figure', figsize=(10, 6))
np.set_printoptions(precision=4, suppress=True)
```

```python
file_path = '/content/drive/MyDrive/MSSP607/Modules/Module4/parsed.csv'
df = pd.read_csv(file_path)
df['mag'] = pd.to_numeric(df['mag'], errors='coerce')
df.head()
```

```python
#a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
place_Japan = df['place'].str.contains('Japan', na = False)
magType_mb = df['magType'] == 'mb'
Japan_mb_earthquake = df[place_Japan & magType_mb]
magnitude = Japan_mb_earthquake['mag']
percentile_95 = np.percentile(magnitude, 95)
percentile_95_formatted = float(percentile_95)
print(percentile_95_formatted)
```

```python
#b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
place_indonesia = df['place'].str.contains('Indonesia', na = False)
indonesia_earthquake = df[place_indonesia]
indonesia_earthquake_count = len(indonesia_earthquake)
tsunami = df['tsunami'] == 1
indonesia_tsunami = df[place_indonesia & tsunami]
indonesia_tsunami_count = len(indonesia_tsunami)
percentage = (indonesia_tsunami_count / indonesia_earthquake_count)
percentage_formatted = float(percentage)
print(percentage_formatted)
```

```python
#c. Get summary statistics for earthquakes in Nevada
place_nevada = df['place'].str.contains('Nevada', na = False)
nevada_earthquake = df[place_nevada]
nevada_earthquake.describe()
```

```python
#d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire. Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico (be sure not to select New Mexico), Japan, Philippines, Indonesia, New Zealand, Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington, California, Russia, Taiwan, Tonga, and the Kermadec Islands).
ring_of_fire = ['Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
                'Guatemala', 'Mexico', 'Japan', 'Philippines', 'Indonesia', 'New Zealand',
                'Antarctic', 'Canada', 'Fiji', 'Alaska', 'Washington', 'California', 'Russia', 'Taiwan',
                'Tonga', 'Kermadec Islands'
]
search_method = '|'.join(ring_of_fire)
is_ring_of_fire = df['place'].str.contains(search_method, na = False, case=False)
is_new_mexico = df['place'].str.contains('New Mexico', na = False, case=False)
df['Ring_of_Fire'] = is_ring_of_fire & ~ is_new_mexico
df[is_new_mexico].head()
df.to_csv('earthquakes_with_ring_of_fire.csv', index=False)
```

```python
#e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
ring_of_fire_count = df['Ring_of_Fire'].sum()
outside = len(df) - ring_of_fire_count
print(ring_of_fire_count)
print(outside)
```

```python
#f. Find the tsunami count along the Ring of Fire
is_ring_of_fire_true = df['Ring_of_Fire'] == True
ring_of_fire_earthquakes = df[is_ring_of_fire_true]
tsunami_count = ring_of_fire_earthquakes['tsunami'].sum()
print(tsunami_count)
```
